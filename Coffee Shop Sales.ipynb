```python
df.drop(columns=['transaction_id', 'transaction_date', 'transaction_time','store_id','product_id','product_detail'],inplace=True)
```

```python
df.info()
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df['Total Price']=df['transaction_qty']* df['unit_price']
df
```

```python
totalamt=df['Total Price'].sum()
print('Total Price:',totalamt)
```

```python
totalqty=df['transaction_qty'].sum()
print('transaction_qty:',totalqty)
```

```python
stores=df['store_location'].value_counts().reset_index()
stores
```

```python
plt.figure(figsize=(4,3))
plt.bar(stores['store_location'], stores['count'],color= 'black',width=0.3)
plt.xlabel('Store Location',color='red')
plt.ylabel('Store count',color='red')
plt.grid(color='red')
plt.show()
```

```python
store_amount=df.groupby('store_location')[['Total Price']].sum().reset_index()
store_amount
```

```python
plt.figure(figsize=(4,3))
plt.bar(store_amount['store_location'],store_amount['Total Price'],color='black',width=0.3)
plt.xlabel('Store Location',color='red')
plt.ylabel('Total Price',color='red')
plt.grid(color='red')
plt.show()
```

```python
amountByProduct1=df.groupby('product_category')[['Total Price']].sum().reset_index()
amountByProduct1
```

```python
plt.figure(figsize=(6,5))
plt.bar(amountByProduct1['product_category'],amountByProduct1['Total Price'],color='red',width=0.5)
plt.xlabel('Product Category',color='red')
plt.ylabel('Total Price',color='red')
plt.grid(color='red')
plt.plot(amountByProduct1['Total Price'],color='black',marker='.')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.pie(amountByProduct1['Total Price'])
plt.legend(bbox_to_anchor=(1.05,1.0),labels=amountByProduct1['product_category'])
plt.show()
```

```python
data1=df.where(df['store_location']=='Astoria')
data1.dropna(inplace=True)
```

```python
amountByProduct2=data1.groupby('product_category')[['transaction_qty','Total Price']].sum().reset_index()
amountByProduct2
```

```python
c=np.arange(len(amountByProduct2['product_category']))
plt.xticks(c,amountByProduct2['product_category'],rotation=90)
plt.bar(c,amountByProduct2['transaction_qty'],width=0.5,label='Quantity',color='red')
plt.bar(c+0.3,amountByProduct2['Total Price'],width=0.5,label='Total Price',color='black')
plt.xlabel('Product Category',color='red')
plt.ylabel('Amount and Quantity',color='red')
plt.legend()
plt.show()
```
